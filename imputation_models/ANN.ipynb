```python
df_time = pd.DataFrame()

df = pd.DataFrame()
df_vars = pd.DataFrame()

df_t = pd.DataFrame()
df_vars_t = pd.DataFrame()
for j in range(5, 6):
  print('Доля пропусков', j)
  print('--------------------------------------')
  file_mis = '/content/drive/MyDrive/датасеты/данные диплом/0'+str(j)+'/data0'+str(j) +'.json'
  file_tar = '/content/drive/MyDrive/датасеты/данные диплом/0'+str(j)+'/data0'+str(j) +'_tar.json'

  data_mis = pd.read_json(file_mis)
  data_tar =  pd.read_json(file_tar)

  for ind in range(300, data_mis.shape[0]):
    start_time = time.time()
    df.loc[ind, '%'] =  j/10

    dfm1 = pd.DataFrame(data_mis.loc[ind].to_dict())
    dft1 = pd.DataFrame(data_tar.loc[ind].to_dict())

    dfm1 = dfm1[sorted(dfm1.columns)].reset_index(drop = True)
    dft1 = dft1[sorted(dfm1.columns)].reset_index(drop = True)

    # dfm = pd.concat([dfm1[int_cols], dft1[cat_ord]], axis=1)
    dfm = pd.concat([dfm1[cat_ord], dft1[int_cols]], axis=1)
    dfm = dfm[sorted(dft1.columns)]

    for i in mis_cat: # mis_cat
        # print(i)
        # print(True)
        y = dft1[i]
        y = pd.get_dummies(y, dummy_na=False, drop_first=False, dtype=int)
        X = dft1.drop(i, axis=1)
        X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.3, random_state=42)


        ann = tf.keras.models.Sequential()
        ann.add(tf.keras.layers.Dense(units=len(y.columns),activation="relu"))
        # ann.add(tf.keras.layers.Dropout(0.5))
        ann.add(tf.keras.layers.Dense(units=6,activation="relu"))
        ann.add(tf.keras.layers.Dense(units=18,activation="relu"))
        ann.add(tf.keras.layers.Dense(units=len(y.columns),activation="softmax"))

        ann.compile(optimizer="adam",loss="categorical_crossentropy",metrics=['f1_score', 'accuracy'])


        ann.fit(X_train, y_train, batch_size=32,epochs = 20, verbose = 0)

        # test_loss = ann.evaluate(x = X_test, y = y_test)
        l = ann.predict(X_test)
        class_pred = np.argmax(l, axis=1)+1

        pred = class_pred[dfm.loc[X_test.index][i].isna() == 1]

        true = dft1.loc[X_test.index][i][dfm.loc[X_test.index][i].isna() == 1]

        t_pred = ann.predict(dft1[dfm[i].isna() == 1].drop(i, axis=1))
        class_pred_t = np.argmax(t_pred, axis=1)+1

        not_imputed = dft1.loc[dft1[dfm[i].isna() != 1].index][i]
        not_imputed = not_imputed.to_list()
        not_imputed.extend(class_pred_t.tolist())

        f1_score = f1(true, pred, average = 'macro')
        ac_score = acc(true, pred)
        # print(ind)
        # print(f1_score, ac_score)

        df.loc[ind, i+'_f1'] = f1_score
        df.loc[ind, i+'_acc'] = ac_score
        df_vars.loc[ind, i] = np.array(not_imputed).var()
        print(df.shape)

    print("--- %s seconds ---" % (time.time() - start_time))
    df_time.loc[ind, 'time'] = time.time() - start_time

    df_t = pd.concat([df_t, df])
    df_t = df_t.drop_duplicates()
    df_vars_t = pd.concat([df_vars_t, df_vars])
    df_vars_t = df_vars_t.drop_duplicates()

    print(df_t.shape)
    print('Обработан датасет №', ind)

    # break
```
