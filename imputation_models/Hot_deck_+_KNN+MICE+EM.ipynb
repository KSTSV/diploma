# hot_deck

```python
from sklearn.impute import KNNImputer
import pandas as pd


df = pd.DataFrame(dfm1)
def hot_deck(df):
  # Initialize KNNImputer
  imputer = KNNImputer(n_neighbors=1)

  # Impute missing values
  imputed_data = imputer.fit_transform(df)

  # Convert back to DataFrame
  imputed_df = pd.DataFrame(imputed_data, columns=df.columns)
  return imputed_df
# imputed_df.isna().sum().sum()
```

```python
df = pd.DataFrame()
df_total = pd.DataFrame()
df_stat = pd.DataFrame()
df_stat_total = pd.DataFrame()
df_vars = pd.DataFrame()
df_vars_full = pd.DataFrame()

import time


for i in range(5, 6):
  print('Доля пропусков', i)
  print('--------------------------------------')
  file_mis = '/content/drive/MyDrive/Диплом/Датасеты/данные диплом/0'+str(i)+'/data0'+str(i) +'_.json'
  file_tar = '/content/drive/MyDrive/Диплом/Датасеты/данные диплом/0'+str(i)+'/data0'+str(i) +'_tar_.json'

  data_mis = pd.read_json(file_mis)
  data_tar =  pd.read_json(file_tar)

  for ind in range(data_mis.shape[0]):
    start_time = time.time()
    df.loc[ind, '%'] = i/10

    df_total.loc[ind, '%'] =  i/10

    dfm1 = pd.DataFrame(data_mis.loc[ind].to_dict())
    dft1 = pd.DataFrame(data_tar.loc[ind].to_dict())

    dfm1 = dfm1[sorted(dfm1.columns)].reset_index(drop = True)
    dft1 = dft1[sorted(dfm1.columns)].reset_index(drop = True)

    imp_data = hot_deck(dfm1).reset_index(drop = True)
    print("--- %s seconds ---" % (time.time() - start_time))

    for var in dfm1.columns:
    # for var in ['agea']:
      # print('-------------------------', var)
      index_mis = list(map(int, dfm1[var][dfm1[var].isna()].index))
      mis_imput =  imp_data[var].loc[index_mis]
      tar_vals = dft1[var].loc[index_mis]
      if len(tar_vals) == 0:
        df[var] = 'full'
        continue
      else:
        if var in int_cols:
          error = rmse(tar_vals, mis_imput)
          df.loc[ind, var] = error
          # print(var, error)
          df_vars.loc[ind, var] = imp_data[var].var()


        if var in ord_cols:
          mis_imput = mis_imput.apply(lambda row: math.ceil(row))
          error = f1(tar_vals, mis_imput, average ='macro')
          df.loc[ind, var] = error
          # print(var, error)
          df_vars.loc[ind, var] = entropy(imp_data[var])

        if var in cat_cols:
          error = f1(tar_vals, mis_imput, average ='macro')
          df.loc[ind, var] = error
          # print(var, error)
          df_vars.loc[ind, var] = entropy(imp_data[var])

  # total metrics
    imputed_data = hot_deck(dfm1).reset_index(drop = True)
    im_int = imputed_data[list(set(miss_vars) & set(int_cols))]
    im_ord = imputed_data[list(set(miss_vars) & set(ord_cols))].apply(np.ceil)
    im_cat = imputed_data[list(set(miss_vars) & set(cat_cols))]
    im_cat = imputed_data[list(set(miss_vars) & set(cat_cols))]

    im_cat_mar = imputed_data[mar_cat]
    im_cat_mcar = imputed_data[mcar_cat]

    im_ord_mar = imputed_data[mar_ord]
    im_ord_mcar = imputed_data[mcar_ord]

    im_int_mar = imputed_data[mar_int]
    im_int_mcar = imputed_data[mcar_int]

    imputedvars_int = np.array(im_int)[np.array(dfm1[miss_vars][int_cols].isna())]
    imputedvars_ord = np.array(im_ord)[np.array(dfm1[miss_vars][ord_cols].isna())]
    imputedvars_cat = np.array(im_cat)[np.array(dfm1[list(set(miss_vars) & set(cat_cols))].isna())]

    mar_imputedvars_cat = np.array(im_cat_mar)[np.array(dfm1[mar_cat].isna())]
    mar_imputedvars_ord = np.ceil(np.array(im_ord_mar)[np.array(dfm1[mar_ord].isna())])
    mar_imputedvars_int = np.array(im_int_mar)[np.array(dfm1[mar_int].isna())]
    mcar_imputedvars_cat = np.array(im_cat_mcar)[np.array(dfm1[mcar_cat].isna())]
    mcar_imputedvars_ord = np.ceil(np.array(im_ord_mcar)[np.array(dfm1[mcar_ord].isna())])
    mcar_imputedvars_int = np.array(im_int_mcar)[np.array(dfm1[mcar_int].isna())]


    TargetTagsvars_int = np.array(dft1[miss_vars][int_cols])[dfm1[miss_vars][int_cols].isna()]
    TargetTagsvars_ord = np.array(dft1[miss_vars][ord_cols])[dfm1[miss_vars][ord_cols].isna()]
    TargetTagsvars_cat = np.array(dft1[list(set(miss_vars) & set(cat_cols))])[dfm1[list(set(miss_vars) & set(cat_cols))].isna()]

    mar_TargetTagsvars_cat = np.array(dft1[mar_cat])[dfm1[mar_cat].isna()]
    mar_TargetTagsvars_ord = np.array(dft1[mar_ord])[dfm1[mar_ord].isna()]
    mar_TargetTagsvars_int = np.array(dft1[mar_int])[dfm1[mar_int].isna()]
    mcar_TargetTagsvars_cat = np.array(dft1[mcar_cat])[dfm1[mcar_cat].isna()]
    mcar_TargetTagsvars_ord = np.array(dft1[mcar_ord])[dfm1[mcar_ord].isna()]
    mcar_TargetTagsvars_int = np.array(dft1[mcar_int])[dfm1[mcar_int].isna()]


    df_total.loc[ind, 'total_rmse_int'] = rmse(TargetTagsvars_int, imputedvars_int)

    # df_total.loc[ind, 'total_rmse_ord'] = rmse(TargetTagsvars_ord, imputedvars_ord )
    df_total.loc[ind, 'total_acc_ord'] = acc(TargetTagsvars_ord, imputedvars_ord )
    df_total.loc[ind, 'total_f1_ord']  = f1(TargetTagsvars_ord, imputedvars_ord, average ='macro' )
    # df_total['total_pr_ord']  = pr(TargetTagsvars_ord, imputedvars_ord, average ='macro')
    # df_total['total_rc_ord'] = rc(TargetTagsvars_ord, imputedvars_ord, average ='macro')

    df_total.loc[ind, 'total_acc_cat'] = acc(TargetTagsvars_cat, imputedvars_cat)
    df_total.loc[ind, 'total_f1_cat'] = f1(TargetTagsvars_cat, imputedvars_cat, average ='macro')
    # df_total['total_pr_cat'] = pr(TargetTagsvars_cat, imputedvars_cat, average ='macro')
    # df_total['total_rc_cat'] = rc(TargetTagsvars_cat, imputedvars_cat, average ='macro')

    #mar
    df_total.loc[ind, 'mar_f1_cat']  = f1(mar_TargetTagsvars_cat, mar_imputedvars_cat, average ='macro' )
    df_total.loc[ind, 'mar_f1_ord']  = f1(mar_TargetTagsvars_ord, mar_imputedvars_ord, average ='macro' )
    df_total.loc[ind, 'mar_rmse_int']  = rmse(mar_TargetTagsvars_int, mar_imputedvars_int)
    #mcar
    df_total.loc[ind, 'mcar_f1_cat']  = f1(mcar_TargetTagsvars_cat, mcar_imputedvars_cat, average ='macro' )
    df_total.loc[ind, 'mcar_f1_ord']  = f1(mcar_TargetTagsvars_ord, mcar_imputedvars_ord, average ='macro' )
    df_total.loc[ind, 'mcar_rmse_int']  = rmse(mcar_TargetTagsvars_int, mcar_imputedvars_int)

    if ind%20 == 0:
      print(f'Загружено {ind} из {data_mis.shape[0]}' )

    # if ind == 50: break

    df_stat = pd.concat([df_stat, df])

    df_stat_total = pd.concat([df_stat_total, df_total])
    df_vars_full = pd.concat([df_vars_full, df_vars])

    df_stat = df_stat.drop_duplicates()
    df_stat_total = df_stat_total.drop_duplicates()

    df_vars_full = df_vars_full.drop_duplicates()

  print(df_stat.shape)
  print(df_stat_total.shape)
  print(df_vars_full.shape)
```

```python
def impute_em(X, max_iter = 30, eps = 1e-08):
    '''(np.array, int, number) -> {str: np.array or int}

    Precondition: max_iter >= 1 and eps > 0

    Return the dictionary with five keys where:
    - Key 'mu' stores the mean estimate of the imputed data.
    - Key 'Sigma' stores the variance estimate of the imputed data.
    - Key 'X_imputed' stores the imputed data that is mutated from X using
      the EM algorithm.
    - Key 'C' stores the np.array that specifies the original missing entries
      of X.
    - Key 'iteration' stores the number of iteration used to compute
      'X_imputed' based on max_iter and eps specified.
    '''
    columns = X.columns
    X = np.array(X)

    nr, nc = X.shape # количество строк и столбцов
    C = np.isnan(X) == False

    # Collect M_i and O_i's
    one_to_nc = np.arange(1, nc + 1, step = 1)
    M = one_to_nc * (C == False) - 1
    O = one_to_nc * C - 1

    # генерация изначльных параметров выборки (среднее и дисперсия)
    Mu = np.nanmean(X, axis = 0)
    observed_rows = np.where(np.isnan(sum(X.T)) == False)[0]
    S = np.cov(X[observed_rows, ].T)
    if np.isnan(S).any():
        S = np.diag(np.nanvar(X, axis = 0))

    # Start updating
    Mu_tilde, S_tilde = {}, {}
    X_tilde = X.copy()
    no_conv = True
    iteration = 0
    while no_conv and iteration < max_iter:
        for i in range(nr):
            S_tilde[i] = np.zeros(nc ** 2).reshape(nc, nc)
            if set(O[i, ]) != set(one_to_nc - 1): # missing component exists
                M_i, O_i = M[i, ][M[i, ] != -1], O[i, ][O[i, ] != -1]
                S_MM = S[np.ix_(M_i, M_i)]
                S_MO = S[np.ix_(M_i, O_i)]
                S_OM = S_MO.T
                S_OO = S[np.ix_(O_i, O_i)]
                Mu_tilde[i] = Mu[np.ix_(M_i)] +\
                    S_MO @ np.linalg.pinv(S_OO) @(X_tilde[i, O_i] - Mu[np.ix_(O_i)])
                X_tilde[i, M_i] = Mu_tilde[i]
                S_MM_O = S_MM - S_MO @ np.linalg.pinv(S_OO) @ S_OM

                S_tilde[i][np.ix_(M_i, M_i)] = S_MM_O
        Mu_new = np.mean(X_tilde, axis = 0)
        S_new = np.cov(X_tilde.T, bias = 1) +\
            reduce(np.add, S_tilde.values()) / nr
        no_conv =\
            np.linalg.norm(Mu - Mu_new) >= eps or\
            np.linalg.norm(S - S_new, ord = 2) >= eps
        Mu = Mu_new
        S = S_new
        iteration += 1
        if iteration % 10 == 0:
          print('Iteration №', iteration)
    X_tilde = pd.DataFrame(X_tilde, columns = columns)
    result = {
        'mu': Mu,
        'Sigma': S,
        'X_imputed': X_tilde,
        'C': C,
        'iteration': iteration
    }

    return result
```
