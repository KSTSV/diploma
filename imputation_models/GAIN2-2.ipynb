```python
def generator(new_x,m):

    inputs = torch.cat(dim = 1, tensors = [new_x,m])  # маска пропусков + данные
    G_h1 = F.leaky_relu(torch.matmul(inputs, G_W1) + G_b1)
    # G_h1 = F.batch_norm(G_h1)
    G_h1_drop = F.dropout(G_h1, p=0.2, training=True)
    G_h2 = F.relu(torch.matmul(G_h1_drop, G_W2) + G_b2)
    # G_h1 = F.batch_norm(G_h2)
    # G_prob = torch.sigmoid(torch.matmul(G_h2, G_W3) + G_b3) # [0,1] normalized Output
    G_prob = torch.sigmoid(torch.matmul(G_h2, G_W3) + G_b3) # [0,1] нормализованный вывод

    return G_prob

#%% 2. Discriminator
def discriminator(new_x, h):

    inputs = torch.cat(dim = 1, tensors = [new_x,h])  # подсказанные значения + данные
    D_h1 = F.relu(torch.matmul(inputs, D_W1) + D_b1)
    D_h1_drop = F.dropout(D_h1, p=0.6, training=True)

    D_h2 = F.relu(torch.matmul(D_h1, D_W2) + D_b2)
    D_logit = torch.matmul(D_h2, D_W3) + D_b3
    D_prob = torch.sigmoid(D_logit)  # [0,1] вероятность - является ли значение реальным

    return D_prob

#%% функции
# генерация выборки из Z
def sample_Z(m, n):
    return np.random.uniform(0., 0.01, size = [m, n])

# генерация мини-батча
def sample_idx(m, n):
    A = np.random.permutation(m)
    idx = A[:n]
    return idx


def discriminator_loss(M, New_X, H):
    # генератор
    G_sample = generator(New_X,M)
    # совмещаем с исходными данными
    Hat_New_X = New_X * M + G_sample * (1-M)

    # дискриминатор
    D_prob = discriminator(Hat_New_X, H)

    # ошибка (функционал)
    D_loss = -torch.mean(M * torch.log(D_prob + 1e-8) + (1-M) * torch.log(1. - D_prob + 1e-8), dtype=torch.float32)
    return D_loss

def generator_loss(X, M, New_X, H):

    # генератор

    G_sample = generator(New_X,M)

    # овмещаем с исходными данными
    Hat_New_X = New_X * M + G_sample * (1-M)

    # дискриминатор
    D_prob = discriminator(Hat_New_X, H)

    #%% Loss
    G_loss1 = -torch.mean((1-M) * torch.log(D_prob + 1e-8), dtype=torch.float32) # ошибка (без ошибки обучения)

    # ошибка (MSE - для интервальных)
    MSE_train_loss1 = torch.mean((M * New_X - M * G_sample)**2, dtype=torch.float32) / torch.mean(M, dtype=torch.float32)
    # # ошибка (accuracy - для категориальных)
    # t = ((M) * X)
    # p = ((M)*G_sample)
    # MSE_train_loss = ((((t-p).abs() <= 0.3) * (M)).sum() /  (M).sum())

    # ошибка (функционал для категориальных)
    # G_loss = G_loss1 + alpha*0.2 * (1/MSE_train_loss) + alpha*0.8 * MSE_train_loss1

    # ошибка (функционал для интервальных)
    G_loss = G_loss1 + alpha* MSE_train_loss1
    # alpha = alpha*0.9

    #%% ТЕСТОВАЯ ОШИБКА
    # для интервальных
    MSE_test_loss = torch.mean(((1-M) * X - (1-M)*G_sample)**2, dtype=torch.float32) / torch.mean(1-M, dtype=torch.float32)

    # для категориальных
    # t = ((1-M) * X)
    # p = ((1-M)*G_sample)
    # MSE_test_loss = ((((t-p).abs() <= 0.3) * (1-M)).sum() /  (1-M).sum())

    return G_loss, MSE_train_loss1, MSE_test_loss

def test_loss(X, M, New_X):

    # генератор
    G_sample = generator(New_X,M)

    #%% MSE для интервальных
    MSE_test_loss = torch.mean(((1-M) * X - (1-M)*G_sample)**2, dtype=torch.float32) / torch.mean(1-M, dtype=torch.float32)
    #%% accuracy для категориальных
    # t = ((1-M) * X)
    # p = ((1-M) * G_sample)

    #%% общая ошибкаи для категориальных
    # MSE_test_loss = ((((t-p).abs() <= 0.3) * (1-M)).sum() /  (1-M).sum())

    return MSE_test_loss, G_sample

def xavier_init(size):
    in_dim = size[0]
    xavier_stddev = 1. / np.sqrt(in_dim / 2.)
    return np.random.normal(size = size, scale = xavier_stddev)


def sample_M(m, n, p):
    A = np.random.uniform(0., 1., size = [m, n])
    B = A > p
    C = 1.*B
    return C
```
